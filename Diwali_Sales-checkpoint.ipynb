```python
ax = sns.countplot(x="Gender",data=df)
```

```python
ax = sns.countplot(x="Gender",data=df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
sales_gen= df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Gender',y='Amount',data=sales_gen)
for bars in ax.containers:
    ax.bar_label(bars)
    
```

```python
sales_age=df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount')

sns.barplot(x='Age Group',y='Amount',data=sales_age,hue='Age Group')
```

```python
ax= sns.countplot(data=df,x="Age Group",hue= "Gender")

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
df.info()
```

```python
sales_state=df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders')

sns.barplot(x='State',y='Orders',data=sales_state,hue='State')
plt.xticks(rotation=90)
plt.show()
```

```python
sales_state=df.groupby(['State'],as_index=False)['Amount'].sum().sort_values(by='Amount')

sns.barplot(x='State',y='Amount',data=sales_state,hue='State')
plt.xticks(rotation=90)
plt.show()
```
