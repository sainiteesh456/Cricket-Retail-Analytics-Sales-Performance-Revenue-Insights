```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
cric['quantity']=cric['quantity'].astype('int')
cric['unit_price_inr']=cric['unit_price_inr'].astype('int')
cric['customer_rating']=cric['customer_rating'].astype('int')
```

```python
cric.info()
```

```python
cric.info()
```

```python
cric[['quantity','unit_price_inr','customer_rating','discount_pct']]=cric[['quantity','unit_price_inr','customer_rating','discount_pct']].astype('int')
```

```python
cric.info()
```

```python
cric['revenue']=cric['revenue'].astype('int')
```

```python
cric.head()
```

```python
cric.reset_index(drop=True,inplace=True)
```

```python
cric['city'] = cric['city'].str.strip().str.title()
```

```python
cric['revenue'] = cric['quantity']*cric['unit_price_inr']*(1-(cric['discount_pct']/100))

cric.to_csv('cric.csv', index=False)
```

```python


city_revenue = cric.groupby('city',as_index=False)['revenue'].sum().sort_values(by='revenue',ascending=False)
plt.figure(figsize=(8,5))
x=sns.barplot(x='city',y='revenue',data=city_revenue,errorbar=None,hue='city')
for bars in x.containers:
    x.bar_label(bars)
plt.xticks(rotation=45)

plt.title("Revenue by city")
plt.xlabel("City")
plt.ylabel("Revenue in INR")


plt.show()
```

```python
purchase_by_channel=cric.groupby('sales_channel',as_index=False)['revenue'].sum().sort_values(by='revenue',ascending=False)
y=sns.barplot(x='sales_channel',y='revenue',data=purchase_by_channel,hue='sales_channel')

for bars in y.containers:
    y.bar_label(bars)
```

```python
equipment_category_vs_quantity=cric.groupby('equipment_category',as_index=False)['revenue'].sum().sort_values(by='revenue')
x=sns.barplot(x='equipment_category',y='revenue',data=equipment_category_vs_quantity,errorbar=None,hue='equipment_category')


plt.title('Equipment sold vs Revenue')
plt.xticks(rotation=50,ha='right')

plt.figure(figsize=(8,5))


plt.show()
```
